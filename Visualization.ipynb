```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px
```

```python
Month =  ['Jan','Feb','Mar','Apr','May','Jun','jul','Aug','Sep','Oct','Nov','Dec']
Sales =[234,354,466,577,345,345,234,345,345,355,677,456]
Profit = [150,200,300,400,200,200,100,300,300,200,300,100]
Qty =  [40,50,10,20,50,40,20,80,90,20,70,90]
```

```python
plt.figure(figsize=(15,5))
plt.plot(Month , Sales , color = 'green' ,  linestyle='dotted' , marker='o',label='SALES')
plt.plot(Month,Profit,color='orange',marker='*',linestyle='dashed',label='PROFIT')
plt.fill_between(Month,Qty,label='Quantity')
for M,S in zip(Month,Sales):
    plt.text(x=M,y=S,s=str(round(S/1000,1))+'K')
plt.legend()
plt.title('Monthly Numbers')
plt.xlabel('Months')
plt.ylabel('KPIs')
plt.gca().set_facecolor('black')
plt.show()
```

```python
Products = ['T-shirt','Trouser','shoes','belt','cap','socks']
Sales = [2342,4553,554,4455,5455,4545]
```

```python
plt.figure(figsize=(15,5))
plt.bar(Products,Sales,color=['red','blue','green','yellow','pink','purple'])
for P,S in zip(Products,Sales):
    plt.text(x=P,y=S,s=S)
plt.title('Product Wise Sales')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()
```
